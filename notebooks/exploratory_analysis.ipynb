# Color-Invariant Saree Design Recognition
## Exploratory Data Analysis, Texture vs Color Analysis & Grad-CAM Explainability

**Objective**: Demonstrate that traditional CNNs overfit to dominant fabric colors, whereas our proposed **Color-Invariant Architecture** extracts robust structural features (zari borders, butta motifs, diamond weaves) achieving high **Color Consistency Scores** across severe chromatic shifts.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2
import torch

# Check compute environment
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Active device: {device}")
```

Inspect the 6 primary categories: Banarasi, Kanjivaram, Ikat, Kalamkari, Bandhani, Paithani.

### 2. Chromatic Perturbations (Color Invariance Test Suite)
Generate canonical color variants (Red, Blue, Green, Yellow tints, Low Saturation, Grayscale) to test structural invariance.

```python
from src.augmentation import generate_color_variants

# Load a sample saree image if available
sample_path = 'src/assets/images/banarasi_sample_1791210617166.jpg'
if os.path.exists(sample_path):
    sample_img = Image.open(sample_path).convert('RGB')
    variants = generate_color_variants(sample_img)
    
    fig, axes = plt.subplots(1, len(variants), figsize=(18, 3))
    for ax, (name, img) in zip(axes, variants.items()):
        ax.imshow(img)
        ax.set_title(name.replace('_', ' ').title(), fontsize=10)
        ax.axis('off')
    plt.show()
```

Compare Model A (Baseline RGB) vs Model B (Color-Invariant).
